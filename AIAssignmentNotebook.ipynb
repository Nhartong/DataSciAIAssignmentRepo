# Prompt 4
Can you please generate code to create bivariate visualizations for the distributions following variables compared to their hue for 'fraud' class with both a countplot and a groupby bar plot showing the percentage of each variable compared to fraud: used_pin_number, used_chip, and online_order? The percentage plots should be stacked to show that it is out of 100 with a coloring for fraud vs non fraudulent and each plot should contain a labled legend. Coloring should also be red for fraud and green for notfradulent

- Microsoft 365 Copilot

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Features to analyze
features = ['used_pin_number', 'used_chip', 'online_order']

# Color scheme
fraud_color = 'red'
nonfraud_color = 'green'

sns.set_style("whitegrid")

for feature in features:

    fig, axes = plt.subplots(1, 2, figsize=(15, 5))

    # =========================================================
    # COUNTPLOT
    # =========================================================

    sns.countplot(
        data=df_clean,
        x=feature,
        hue='fraud',
        palette={
            0: nonfraud_color,
            1: fraud_color
        },
        ax=axes[0]
    )

    axes[0].set_title(f'{feature} vs Fraud Count')
    axes[0].set_xlabel(feature)
    axes[0].set_ylabel('Count')

    # Custom legend labels
    handles, labels = axes[0].get_legend_handles_labels()
    axes[0].legend(
        handles,
        ['Not Fraudulent', 'Fraudulent'],
        title='Transaction Type'
    )

    # =========================================================
    # 100% STACKED PERCENTAGE BAR CHART
    # =========================================================

    # Calculate percentages within each category
    pct_df = pd.crosstab(
        df_clean[feature],
        df_clean['fraud'],
        normalize='index'
    ) * 100

    # Ensure both columns exist
    if 0 not in pct_df.columns:
        pct_df[0] = 0
    if 1 not in pct_df.columns:
        pct_df[1] = 0

    pct_df = pct_df[[0, 1]]

    # Plot stacked bars
    pct_df.plot(
        kind='bar',
        stacked=True,
        color=[nonfraud_color, fraud_color],
        ax=axes[1],
        width=0.7
    )

    axes[1].set_title(f'Fraud Percentage by {feature}')
    axes[1].set_xlabel(feature)
    axes[1].set_ylabel('Percentage (%)')
    axes[1].set_ylim(0, 100)

    axes[1].legend(
        ['Not Fraudulent', 'Fraudulent'],
        title='Transaction Type',
        loc='upper right'
    )

    # Add percentage labels inside bars
    for container in axes[1].containers:
        labels = [
            f'{value:.1f}%' if value >= 3 else ''
            for value in container.datavalues
        ]

        axes[1].bar_label(
            container,
            labels=labels,
            label_type='center',
            color='white',
            fontsize=9,
            fontweight='bold'
        )

    plt.tight_layout()
    plt.show()
```

# Prompt 6
The next goal is to predict fraudulant transactions. Can you please create new datasets where you select fraud as the target variable and split the data into training and testing sets where you use 80% of the original cleaned data to try to predict the last 20? For now I just need the code to split the data up and display a little bit of each part.

- Microsoft 365 Copilot

```python
# ============================================================
# Define Features (X) and Target (y)
# ============================================================

X = df_clean.drop('fraud', axis=1)
y = df_clean['fraud']

# Display shapes
print("Feature Matrix Shape:", X.shape)
print("Target Vector Shape:", y.shape)

# ============================================================
# Split into Training and Testing Sets
# ============================================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,      # 20% testing
    random_state=42,     # reproducible results
    stratify=y           # preserve fraud distribution
)

# ============================================================
# Display Shapes
# ============================================================

print("\nTraining Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)
print("Training Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)

# ============================================================
# Display First Few Rows
# ============================================================

print("\n========== X_train ==========")
display(X_train.head())

print("\n========== X_test ==========")
display(X_test.head())

print("\n========== y_train ==========")
display(y_train.head())

print("\n========== y_test ==========")
display(y_test.head())
```

# Prompt 7
Can you now create and test (using a confusion matrix) a few models. The first should be one that always predicts no fraud, The next should be a k-NN classifier with 3 n_neighbors and no scaler. These models should then be compared on accuracy, percision, recall score, and f1 score through a labled bar plot. 

- Microsoft 365 Copilot

```python
# ============================================================
# IMPORTS
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.dummy import DummyClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# ============================================================
# MODEL 1: ALWAYS PREDICT NO FRAUD
# ============================================================

baseline_model = DummyClassifier(
    strategy='constant',
    constant=0
)

baseline_model.fit(X_train, y_train)

y_pred_baseline = baseline_model.predict(X_test)

# ============================================================
# MODEL 2: K-NN (k=3, NO SCALING)
# ============================================================

knn_model = KNeighborsClassifier(
    n_neighbors=3
)

knn_model.fit(X_train, y_train)

y_pred_knn = knn_model.predict(X_test)

# ============================================================
# CONFUSION MATRICES
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Baseline Model
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_baseline,
    cmap='Greens',
    ax=axes[0]
)

axes[0].set_title('Always Predict No Fraud')

# k-NN Model
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_knn,
    cmap='Blues',
    ax=axes[1]
)

axes[1].set_title('k-NN Classifier (k=3)')

plt.tight_layout()
plt.show()

# ============================================================
# CALCULATE PERFORMANCE METRICS
# ============================================================

results = pd.DataFrame({
    'Model': ['Always No Fraud', 'k-NN (k=3)'],

    'Accuracy': [
        accuracy_score(y_test, y_pred_baseline),
        accuracy_score(y_test, y_pred_knn)
    ],

    'Precision': [
        precision_score(y_test, y_pred_baseline, zero_division=0),
        precision_score(y_test, y_pred_knn, zero_division=0)
    ],

    'Recall': [
        recall_score(y_test, y_pred_baseline, zero_division=0),
        recall_score(y_test, y_pred_knn, zero_division=0)
    ],

    'F1 Score': [
        f1_score(y_test, y_pred_baseline, zero_division=0),
        f1_score(y_test, y_pred_knn, zero_division=0)
    ]
})

print("Performance Metrics")
display(results.round(4))

# ============================================================
# BAR PLOT COMPARISON
# ============================================================

metrics_long = results.melt(
    id_vars='Model',
    var_name='Metric',
    value_name='Score'
)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=metrics_long,
    x='Metric',
    y='Score',
    hue='Model',
    palette=['green', 'blue']
)

# Add labels to bars
for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.3f',
        padding=3
    )

plt.title('Model Performance Comparison')
plt.xlabel('')
plt.ylabel('Score')
plt.ylim(0, 1.05)

plt.legend(title='Model')
plt.tight_layout()
plt.show()
```
